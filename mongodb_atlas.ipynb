````python
from pymongo import MongoClient
````

````python
client = MongoClient("mongodb+srv://<user>:<password>@<cluster>.mongodb.net/test?retryWrites=true")
````

````python
db = client.get_database('student_db')
````

````python
records = db.student_records
````

## Count documents

````python
records.count_documents({})
````

## Create new document

````python
new_student = {
    'name': 'ram',
    'roll_no': 321,
    'branch': 'it'
}
````

````python
records.insert_one(new_student)
````

````python
new_students = [
    {
        'name': 'alex',
        'roll_no': 320,
        'branch': 'it'
    },
    {
        'name': 'john',
        'roll_no': 30,
        'branch': 'ece'
    }
]
````

````python
records.insert_many(new_students)
````

## Find documents

````python
list(records.find())
````

````python
records.find_one({'roll_no': 123})
````

## Update documents

````python
student_updates = {
    'name': 'Nikhil'
}
````

````python
records.update_one({'roll_no': 123}, {'$set': student_updates})
````

## Delete documents

````python
records.delete_one({'roll_no': 123})
````
